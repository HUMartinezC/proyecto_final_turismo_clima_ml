## Decisiones técnicas

- S3 se usa como repositorio único por capas.
- Glue actúa como catálogo, no como motor de jobs.
- Athena valida y consulta los Parquet publicados.
- Kafka no se implementa porque las fuentes son batch/históricas y no existe flujo de eventos continuo.
- DocumentDB/MongoDB no se implementa porque la trazabilidad documental queda cubierta en raw/bronze y no se requieren consultas documentales para el modelado.
